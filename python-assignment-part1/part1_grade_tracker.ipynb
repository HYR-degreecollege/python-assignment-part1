```python
# Data
raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma",       "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ",    "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA",       "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ",    "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]
```

```python
# Data Cleaning & Storage
cleaned_students = []

for student in raw_students:
    
    name = student["name"].strip().title()
    roll = int(student["roll"])
    marks = list(map(int, student["marks_str"].split(", ")))
    
    is_valid = all(word.isalpha() for word in name.split())
    
    cleaned_students.append({
        "name": name,
        "roll": roll,
        "marks": marks,
        "valid": is_valid
    })
```

```python
# Validation + Profile Display
for student in cleaned_students:
    
    name = student["name"]
    roll = student["roll"]
    marks = student["marks"]
    is_valid = student["valid"]
    
    # Validation output
    if is_valid:
        print(f"{name} : ✓ Valid name")
    else:
        print(f"{name} : ✗ Invalid name")
    
    # Profile display
    print("================================")
    print(f"Student : {name}")
    print(f"Roll No : {roll}")
    print(f"Marks   : {marks}")
    print("================================\n")
```

```python
# Special Output applied on  (Roll 103)
for student in cleaned_students:
    if student["roll"] == 103:
        name = student["name"]
        
        print("Special Output for Roll 103:")
        print("UPPERCASE :", name.upper())
        print("lowercase :", name.lower())
```

```python
student_name = "Ayesha Sharma"
subjects = ["Math", "Physics", "CS", "English", "Chemistry"]
marks = [88, 72, 95, 60, 78]
```

```python
# Q1: Subject-wise Grade Display
print(f"Student: {student_name}\n")

for subject, mark in zip(subjects, marks):
    
    if 90 <= mark <= 100:
        grade = "A+"
    elif 80 <= mark <= 89:
        grade = "A"
    elif 70 <= mark <= 79:
        grade = "B"
    elif 60 <= mark <= 69:
        grade = "C"
    else:
        grade = "F"
    
    print(f"{subject:<10} : {mark} -> Grade {grade}")
```

```python
# Q2: Total, Average, Highest, Lowest
# Total & Average
total_marks = sum(marks)
average_marks = round(total_marks / len(marks), 2)

# Highest
max_mark = max(marks)
highest_subject = subjects[marks.index(max_mark)]

# Lowest
min_mark = min(marks)
lowest_subject = subjects[marks.index(min_mark)]

# Output
print(f"\nStudent: {student_name}\n")
print(f"Total Marks           : {total_marks}")
print(f"Average Marks         : {average_marks}")
print(f"Highest Scoring       : {highest_subject} ({max_mark})")
print(f"Lowest Scoring        : {lowest_subject} ({min_mark})")
```

```python
# Q3: While Loop (Marks Entry System)
print("\nEnter new subjects and marks (type 'done' to stop)\n")

# Simulated inputs (for notebook)
inputs = [
    ("Biology", "85"),
    ("History", "abc"),
    ("Geography", "105"),
    ("Civics", "75"),
    ("done", "")
]

i = 0
new_subject_count = 0

while True:
    
    subject = inputs[i][0]
    print(f"Enter subject name: {subject}")
    
    if subject.lower() == "done":
        break
    
    mark_input = inputs[i][1]
    print(f"Enter marks (0-100): {mark_input}")
    
    i += 1
    
    # Validation
    if not mark_input.isdigit():
        print("⚠ Invalid input! Marks must be numeric.\n")
        continue
    
    mark = int(mark_input)
    
    if mark < 0 or mark > 100:
        print("⚠ Invalid input! Marks must be between 0 and 100.\n")
        continue
    
    subjects.append(subject)
    marks.append(mark)
    new_subject_count += 1
    
    print("✅ Subject added successfully!\n")

# Final Summary
total_marks = sum(marks)
average_marks = round(total_marks / len(marks), 2)

print("\n===== Final Summary =====")
print(f"New subjects added : {new_subject_count}")
print(f"Updated Average    : {average_marks}")
```

```python
#
#
# Task 3 -Class Performance Summary
#
#
```

```python
class_data = [
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]
```

```python
# Q1: Average + Pass/Fail
print("===== Class Performance Summary =====\n")

for name, marks in class_data:
    
    avg = round(sum(marks) / len(marks), 2)
    
    if avg >= 60:
        result = "Pass"
    else:
        result = "Fail"
    
    print(f"Student : {name}")
    print(f"Average : {avg}")
    print(f"Result  : {result}")
    print("----------------------------------")
```

```python
# Q2: Formatted Table
print("Name              | Average | Status")
print("----------------------------------------")

for name, marks in class_data:
    
    avg = round(sum(marks) / len(marks), 2)
    
    if avg >= 60:
        status = "Pass"
    else:
        status = "Fail"
    
    print(f"{name:<18} | {avg:6.2f} | {status}")
```

```python
# Q3: Summary (Pass/Fail + Topper + Class Avg)
pass_count = 0
fail_count = 0
averages = []

topper_name = ""
topper_avg = 0

for name, marks in class_data:
    
    avg = round(sum(marks) / len(marks), 2)
    averages.append(avg)
    
    if avg >= 60:
        pass_count += 1
    else:
        fail_count += 1
    
    if avg > topper_avg:
        topper_avg = avg
        topper_name = name

# Class average
class_avg = round(sum(averages) / len(averages), 2)

print("\n===== Summary =====")
print(f"Passed Students : {pass_count}")
print(f"Failed Students : {fail_count}")
print(f"Class Topper    : {topper_name} ({topper_avg})")
print(f"Class Average   : {class_avg}")
```

```python
# Q-1 Given Data + Step 1 (Strip)
essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

# Step 1
clean_essay = essay.strip()

print("Step 1 — Clean Essay:")
print(clean_essay)
```

```python
# Q-2 Step 2 (Title Case)
title_essay = clean_essay.title()

print("\n 2 — Title Case:")
print(title_essay)
```

```python
# Q-3 Step 3 (Count 'python')
count_python = clean_essay.count("python")

print("\n 3 — Count of 'python':")
print(count_python)
```

```python
# Q-4  (Replace)
replaced_essay = clean_essay.replace("python", "Python 🐍")

print("\n 4 — Replaced Text:")
print(replaced_essay)
```

```python
# Q-5 Split Sentences
sentences = clean_essay.split(". ")

print("\n Sentence List:")
print(sentences)
```

```python
# Q-6 Numbered Sentences
print("\n  Numbered Sentences:")

for i, sentence in enumerate(sentences, start=1):
    
    sentence = sentence.strip()  # extra safety
    
    if not sentence.endswith("."):
        sentence += "."
    
    print(f"{i}. {sentence}")
```
